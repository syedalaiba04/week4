```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
print("=" * 60)
print("dataset 1: pak ecommerce")
print("=" * 60)
ecom = pd.read_csv("pakistani_ecommerce_dataset.csv")
print(ecom.head())
print(ecom.info())
```

QUESTION:1 MOST POPULAR PRODUCT CATEGORY

```python
category_counts=ecom["product_category"].value_counts()
print("\nQ1:orders per product category:")
print(category_counts)
print("most popular category:", category_counts.idxmax(),
      f"({category_counts.max()}orders)")
```

```python
city_counts = ecom['city'].value_counts()
print('\nQ3: order per city:')
print(city_counts)
print("city with most orders:", city_counts.idxmax(),
      f"({city_counts.max()} orders)")
```

QUESTION4: MOST USED PAYMENT METHOD

```python
pay_counts = ecom['payment_method'].value_counts()
print("\nQ4: payment method usage:")
print(pay_counts)
print("most used payment method:" , pay_counts.idxmax(),
      f"({pay_counts.max()} orders)")
```

QUESTION5: BAR CHART

```python
avg_city= ecom.groupby("city")["order_value"].mean()
categories= avg_city.index
values = avg_city.values
plt.bar(categories,values,color='violet', width=0.5)
plt.xlabel("city")
plt.ylabel("average order value")
plt.title("average order value by city")
plt.show()
```

DATA SET 2: STUDENT LIFESTYLE AND STUDY HABITS

```python
stu= pd.read_csv("student_lifestyle_dataset.csv")
print(stu.head())
print(stu.info())
```

QUESTION1: AVERAGE NUMBER OF STUDY HOURS

```python
avg_study=stu["study_hours"].mean()
print(f"\nQ1: average study hours:{avg_study:.2f}")
```

QUESTION2: WHO SLEEPS MORE ON AVERAGE-MALES OR FEMALES

```python
gender_sleep= stu.groupby("gender")["sleep_hours"].mean()
print("\nQ2:average sleep hours by gender:")
print(gender_sleep.round(2))
print("sleeps more on average:", gender_sleep.idxmax())
```

QUESTION3: RELATIONSHIP BETWEEN SCREEN TIME AND SLEEP HOURS

```python
stu['screen_time'].corr(stu['sleep_hours'])
```

QUESTION4: HISTOGRAM

```python
plt.hist(stu["daily_transport_cost"],bins=10, color= 'pink', edgecolor='red', alpha=0.7)
plt.xlabel("daily transport cost (rs.)")
plt.ylabel("number of students")
plt.title("historgam")
plt.show()
```

QUESTION5: COMPARE AVERAGE INTERNET USAGE BY GENDER

```python
internet_by_gender = stu.groupby("gender")["internet_usage"].mean()
print(internet_by_gender)
print("higher average internet usage:", internet_by_gender.idxmax())
```
