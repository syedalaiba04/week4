```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("ride_hailing_dataset.csv")
```

QUESTION1: AVERAGE FARE BY SERVICE TYPE

```python
avg_fare = df.groupby('service_type')['fare'].mean()
print("Q1: average fare by service type:\n", avg_fare)
```

QUESTION2: CITY WITH MOST TRIPS

```python
city_counts = df['city'].value_counts()
print("nQ2: trips per city:\n", city_counts)
print("city with most trips:", city_counts.idxmax())
    
```

QUESTION3: AVERAGE DISTANCE PER TRIP

```python
print("\Q3: average distance per trip:", df['distance_km'].mean())
```

QUESTION4: COMPARE TRIP RATINGS BY GENDER

```python
rating_by_gender= df.groupby('gender')['trip_rating'].mean()
print("\nQ4: average trip rating by gender:\n", rating_by_gender)
```

QUESTION5: BAR CHART

```python
plt.bar(avg_fare.index, avg_fare.values, color='green', width =0.5)
plt.xlabel('service type')
plt.ylabel('average fare ')
plt.title('avg fare by service type')
plt.show()
```
