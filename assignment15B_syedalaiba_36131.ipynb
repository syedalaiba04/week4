```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv('student_tea_snacks.csv')
```

QUESTION1: MOST PREFERRED TEA TYPE

```python
tea_type_counts= df['tea_type'].value_counts()
print('Q2: tea preferrance:\n', tea_type_counts)
print("most preferred tea:", tea_type_counts.idxmax())
```

QUESTION2:AVG NO OF CUPS PER GENDER

```python
print(df.groupby('gender')['cups_per_day'].mean())
```

QUESTION3: BAR CHART OF SNACK CHOICES

```python
snacks = df['snack_choice'].value_counts()
plt.bar(snacks.index, snacks.values, color='blue', width=0.5)
plt.xlabel('Snack')
plt.ylabel('Number of Students')
plt.title('Snack Choices')
plt.show()
```

QUESTION4: TEA PREFERRANCE

```python
print(df.groupby(['age_group', 'tea_type'])['student_id'].count().unstack())
```

QUESTION5: NO OF STUDENTS WHO DONT DRINK TEA

```python
print((df['prefers_tea'] == 'No').sum())
```
